```python
#series creation
s=pd.Series([1,2,3,4,5])
print(s)
print(type(s))

#index
print(s[2])
           
```

```python
#custom label
s=pd.Series([12,28,34,49,50],index=["A","B","C","D","E"])
print(s)
print(s["C"])
print(s.index)
```

```python
#series 
s1=pd.Series([1,2,3,4,5])
s2=pd.Series([10,20,30,40,50])

print(s1*s2)
s1[0]=100
changed_s1=s1.drop(0)

print(s1)
print(changed_s1)

```

```python
#Dataframe
#Creating from dictionary
info={
    "Name": ["Bidisha","John","Bob"],
    "Age":[23,34,56],
    "CGPA": [9.00,8.7,7.5]
    
}

df=pd.DataFrame(info)
print(df)

print(df.index)
print(df.columns)
```

```python
#creating from list of lists
df=pd.DataFrame([["Joy",23],["Bob",56],["Adam",10]] , columns=["Name","Age"])
print(df)
```

```python
#creating from numpy arrays
import numpy as np
np_arr=np.array([[1,2,3],[4,5,6]])
df=pd.DataFrame(np_arr, columns=["A","B","C"])
print(df)
```
