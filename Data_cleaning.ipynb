```python
import pandas as pd
```

```python
#Handle missing data

df=pd.read_csv("raw_data.csv")
df.isnull()
df.isna()
df.isnull().sum()
df.dropna() #row drop containing null values
df.dropna(axis=1) #column drop
mean_age=df["age"].mean()
df.fillna(100)
df.fillna(mean_age)
cleaned_data=df.copy()
mean_age=cleaned_data["age"].mean()
cleaned_data["age"]=cleaned_data["age"].fillna(mean_age)

#cleaned_data
df.ffill() #forward fill

df.bfill()  #backword fill
```

```python
#handle duplicate data

df.duplicated()
df["country"].duplicated() #checking for one column
df[["country","name"]].duplicated() #checking duplicate values for a set of column
df2=df.copy()
df.drop_duplicates(inplace=True)
```

```python
#Data types
df.dtypes
df2=df.copy()
df2 = df2.fillna(0)
df2["age"]=df2["age"].astype("int64")
df2.dtypes
```

```python
#handle strings
df["gender"].str.lower()
df["gender"].str.upper()
df["gender"].str.capitalize()
df["name"].str.split(" ")
df["country"].str.contains("Usa",case=False)
```
